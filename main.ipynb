```python
import sounddevice as sd
from scipy.io.wavfile import write
from IPython.display import Audio
from datetime import datetime

fs = 44100
seconds = 5

print("Gravando...")
audio = sd.rec(int(seconds * fs), samplerate=fs, channels=1)
sd.wait()

timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
file_name = f"audio_{timestamp}.wav"

write(file_name, fs, audio)
print(f"Salvo como {file_name}")

Audio(file_name)
```

```python
pip install openai-whisper
```

```python
import whisper

model = whisper.load_model("small")

result = model.transcribe(file_name, fp16=False, language="pt")
transcription = result["text"]

print(transcription)
```

```python
import openai

# Configura a chave de API da OpenAI usando a variável de ambiente 'OPENAI_API_KEY'
openai.api_key = os.environ.get('OPENAI_API_KEY')

response = openai.ChatCompletion.create(
    model="gpt-4",
    messages=[ { "role": "user", "content": transcription } ]
)

# Obtém a resposta gerada pelo ChatGPT
chatgpt_response = response.choices[0].message.content
print(chatgpt_response)
```

```python
from gtts import  gTTS

gtts_object = gTTS(text=chatgpt_response, lang=language, slow=False)

response_audio = "/content/response_audio.wav"
gtts_object.save(response_audio)

display(Audio(response_audio, autoplay=True))
```
